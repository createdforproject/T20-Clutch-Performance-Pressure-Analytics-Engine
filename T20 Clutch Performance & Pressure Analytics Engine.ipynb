```python
db.query("""
create or replace view main_table as
select b.*,
m.* exclude(match_id),
c.* exclude(comp_id)
from ball_by_ball b 
inner join matches m on b.match_id=m.match_id
inner join competitions c on c.comp_id=m.comp_id
WHERE LOWER(comp_name) LIKE '%indian premier league%'
   OR LOWER(comp_name) LIKE '%ipl%'
   OR (
       home_team IN ('India','Australia','England','South Africa','New Zealand','Pakistan','West Indies','Sri Lanka','Bangladesh','Afghanistan','Ireland','Zimbabwe')
       AND away_team IN ('India','Australia','England','South Africa','New Zealand','Pakistan','West Indies','Sri Lanka','Bangladesh','Afghanistan','Ireland','Zimbabwe')
   );

""").df()
```

```python
db.query("""
SELECT DISTINCT home_team AS team FROM main_table
UNION
SELECT DISTINCT away_team AS team FROM main_table
ORDER BY team
""").show()
```

```python
db.query("""
with calculated_table as(
select striker_name,non_striker_name,
count(runs) as total_balls_played,
sum(runs) as total_runs,
count(distinct match_id) as total_matches,

from main_table
group by striker_name,non_striker_name

)
select striker_name,
non_striker_name,
total_runs,
total_balls_played,
round(total_runs*100.0/total_balls_played,2) as strike_rate,
total_matches
from calculated_table
order by total_balls_played desc,total_runs desc

""").df()
```

```python
db.query("""
with pace_table as(
select count(distinct match_id) as total_matches,
striker_name,
non_striker_name,
count(runs) as total_balls_played_pace,
sum(runs) as total_runs_scored_pace,
round(sum(case when runs=0 then 1 else 0 end)*100.0/count(runs),2) as dot_percentage_pace,
round(sum(case when is_four='True'or is_six='True' then 1 else 0 end)*100.0/count(runs),2) as boundary_percentage_pace,
round(sum(case when is_wicket='True' then 1 else 0 end)*100.0/count(distinct match_id), 2) as dismissal_percentage_pace
from main_table
where pace_or_spin='Pace'
group by striker_name,non_striker_name
order by total_balls_played_pace desc
),
spin_table as(
select count(distinct match_id) as total_matches,
striker_name,
non_striker_name,
count(runs) as total_balls_played_spin,
sum(runs) as total_runs_scored_spin,
round(sum(case when runs=0 then 1 else 0 end)*100.0/count(runs),2) as dot_percentage_spin,
round(sum(case when is_four='True'or is_six='True' then 1 else 0 end)*100.0/count(runs),2) as boundary_percentage_spin,
round(sum(case when is_wicket='True' then 1 else 0 end)*100.0/count(distinct match_id), 2) as dismissal_percentage_spin
from main_table
where pace_or_spin='Spin'
group by striker_name,non_striker_name
order by total_balls_played_spin desc
),
calculated_table as(
select p.*,
s.* exclude(striker_name,non_striker_name,total_matches)
from pace_table p 
full outer join spin_table s on p.striker_name=s.striker_name and p.non_striker_name=s.non_striker_name
)
select * exclude(striker_name,non_striker_name),round((total_runs_scored_pace*100.0/total_balls_played_pace),2) as pace_strike_rate,
round((total_runs_scored_spin*100.0/total_balls_played_spin),2) as spin_strike_rate
from calculated_table
limit 10
""").show()
```

```python
db.query("""
with match_runrate as(
select match_id,
round((sum(runs)*6.0)/count(runs),2) as run_rate
from main_table
WHERE COALESCE(LOWER(extras_type), '') NOT LIKE '%wide%'
group by match_id
),
match_player_rr as(
select match_id,
striker_name,
count(runs) as total_balls,
sum(runs) as total_runs,
round(sum(runs)*100.0/count(runs),2) as strike_rate,
round(sum(case when runs=0 then 1 else 0 end)*100.0/count(runs),2) as dot_ball_pct,
round(sum(case when is_four='True' or is_six='True' then 1 else 0 end)*100.0/count(runs),2) as boundary_pct,
round(sum(case when runs=1 or runs=2 or runs=3 then 1 else 0 end)*100.0/count(runs)) as non_boundary_pct,
round(sum(runs)*6.0/count(runs),2) as player_runrate
from main_table
WHERE COALESCE(LOWER(extras_type), '') NOT LIKE '%wide%'
group by striker_name,match_id
order by total_balls desc
),
calculated_table as(
select mp.*  ,
mr.* exclude(match_id)
from match_player_rr mp
inner join match_runrate mr on mr.match_id=mp.match_id
order by run_rate desc
)
select striker_name,
count(distinct match_id) as total_matches,
round(median(total_balls),2) as median_total_balls,
round(median(total_runs),2) as median_total_runs,
round(median(strike_rate),2) as median_strike_rate,
round(median(run_rate),2) as median_match_run_rate,
round(median(player_runrate),2) as median_player_runrate,
round(median(player_runrate-run_rate),2) as median_net_diff
from calculated_table
group by striker_name
having total_matches>=10 and median_total_balls>10
order by median_net_diff asc
""").df()
```

```python
db.query("""
with player_table as(
select match_id,
innings,
striker_name,
sum(runs) as player_runs,
count(runs) as total_balls_faced,
round(sum(runs)*100.0/count(runs),2) as strike_rate,
sum(case when runs=0 then 1 else 0 end) as dot_balls,
sum(case when is_four='True' or is_six='True' then 1 else 0 end) as player_boundary_balls,
round(sum(case when runs=1 or runs=2 or runs=3 then 1 else 0 end)*100.0/count(runs)) as non_boundary_pct,
round(sum(runs)*6.0/count(runs),2) as player_runrate
from main_table
WHERE COALESCE(LOWER(extras_type), '') NOT LIKE '%wide%' and (innings=1 or innings=2) and 
(first_inn_revised_overs is null and second_inn_revised_overs is null)
group by match_id,innings,striker_name
),
team_table as(
select match_id,
innings,
count(case when coalesce(lower(extras_type),'') not like '%wide%' then runs else null end) as total_balls_in_match,
sum(runs+coalesce(extras1_runs,0)+coalesce(extras_2_runs,0)) as team_runs,
sum(case when runs=0 then 1 else 0 end) as total_match_dot_balls,
sum(case when is_four='True' or is_six='True' then 1 else 0 end) as total_boundaries_scored,
sum(case when is_wicket='True' then 1 else 0 end) as total_wickets,
sum(case when is_wicket='True' and overid<10 then 1 else 0 end) as early_wickets,
lag(sum(runs+coalesce(extras1_runs,0)+coalesce(extras_2_runs,0))) over(partition by match_id order by match_id,innings asc) as first_innings_target,
lag(count(case when coalesce(lower(extras_type),'') not like '%wide%' then runs else null end)) over(partition by match_id order by match_id,innings asc) as first_inning_ball_count
from main_table
where (innings=1 or innings=2) and (first_inn_revised_overs is null and second_inn_revised_overs is null)
group by match_id,innings
),
combined_table as(
select pt.striker_name,
pt.match_id,
pt.innings,
pt.player_runs,
pt.total_balls_faced,
pt.strike_rate,
pt.player_boundary_balls,
pt.non_boundary_pct,
tt.team_runs,
round((pt.player_runs*100.0/tt.team_runs),2) as player_run_contribution_pct,
pt.dot_balls as player_dot_ball,
tt.total_match_dot_balls,
round((pt.dot_balls*100.0/nullif(pt.total_balls_faced,0)),2) as player_dotball_pct,
round((tt.total_match_dot_balls-pt.dot_balls)*100.0/nullif((tt.total_balls_in_match-pt.total_balls_faced),0),2) as team_dot_pct,
tt.early_wickets,
(case when (pt.innings=2 and (first_innings_target*6.0/nullif(120.0,0))>8 and 
((tt.team_runs-pt.player_runs)*100.0/nullif((tt.total_balls_in_match-pt.total_balls_faced),0))<(0.80*(first_innings_target*100.0/nullif(120,0))))
 or (pt.innings=1 and early_wickets>=4) then 'pressure' else 'non pressure' end) as situation
from player_table pt 
inner join team_table tt on pt.match_id=tt.match_id and pt.innings=tt.innings
where pt.total_balls_faced>=10
),
calculated_table as(
select striker_name,
sum(case when situation='pressure' then 1 else 0 end) as pressure_situation_count,
count(distinct match_id) as total_matches,
round(sum(case when situation='pressure' then 1 else 0 end)*100.0/count(distinct match_id),2) as pct_pressure_situation, 
round(median(case when situation='pressure' then strike_rate else null end),2) as median_pressure_strike_rate,
round(median(case when situation='non pressure' then strike_rate else null end),2) as median_non_pressure_strike_rate,
round(median(case when situation='pressure' then player_dotball_pct else null end ),2) as median_pressure_dotball_pct,
round(median(case when situation='non pressure' then player_dotball_pct else null end),2) as median_non_pressure_dotball_pct
from combined_table
group by striker_name
having pressure_situation_count>=8
),
final_table as(
select striker_name,
pressure_situation_count,
median_pressure_strike_rate,
median_non_pressure_strike_rate,
round((100-(median_pressure_strike_rate/nullif(median_non_pressure_strike_rate,0))*100.0),2) as strike_rate_dip,
round((median_pressure_dotball_pct-median_non_pressure_dotball_pct),2) as by_how_much_pressure_increase_dotballs
from calculated_table
where total_matches>=30 
),
output_table as(
select * ,
round(((median_pressure_strike_rate / nullif(median(median_pressure_strike_rate) over(), 0)) + 
(median_pressure_strike_rate / nullif(median_non_pressure_strike_rate, 0))) / 2.0, 3) as clutch_index
from final_table
where strike_rate_dip<10 
order by round(((median_pressure_strike_rate / nullif(median(median_pressure_strike_rate) over(), 0)) + 
(median_pressure_strike_rate / nullif(median_non_pressure_strike_rate, 0))) / 2.0, 3) desc
),
situation_table as(
SELECT situation, COUNT(*) AS innings_count, ROUND(MEDIAN(strike_rate), 2) AS median_sr, ROUND(MEDIAN(player_dotball_pct), 2) AS median_dot_pct
FROM combined_table
GROUP BY situation
)
select * 
from situation_table
limit 10
""").show()
```

```python
db.query("""
with player_table as(
select match_id,
innings,
bowler_name,
(sum(runs)+sum(case when coalesce(lower(extras_type),'') in ('wide','other','overstep') then extras1_runs else 0 end)+
sum(case when coalesce(lower(extras_type),'') in ('wide','other','overstep') then extras_2_runs else 0 end)) as total_runs_bowler_conceeded,
sum(case when coalesce(lower(extras_type),'') not in ('wide','other','overstep') then 1 else 0 end) as total_balls_bowled,
sum(case when coalesce(lower(dismissal_type),'') in ('caught','bowled','hit wicket','lbw','stump') then 1 else 0 end) as total_bowler_wicket,
(sum(case when is_four='True' then 1 else 0 end)+sum(case when is_six='True' then 1 else 0 end)) as total_boundary_count,
sum(case when runs=0 and coalesce(lower(extras_type),'') not in ('wide','other','overstep') then 1 else 0 end) as dot_balls_bowled 
from main_table
where (innings=1 or innings=2) and (first_inn_revised_overs is null and second_inn_revised_overs is null)
group by match_id,innings,bowler_name
),
team_table as(
select match_id,
innings,
count(case when coalesce(lower(extras_type),'') not in ('wide','other','overstep') then runs else null end) as total_balls_in_match,
sum(runs+coalesce(extras1_runs,0)+coalesce(extras_2_runs,0)) as team_runs,
sum(case when runs=0 then 1 else 0 end) as total_match_dot_balls,
sum(case when is_four='True' or is_six='True' then 1 else 0 end) as total_boundaries_scored,
sum(case when is_wicket='True' then 1 else 0 end) as total_wickets,
sum(case when is_wicket='True' and overid<10 then 1 else 0 end) as early_wickets,
lag(sum(runs+coalesce(extras1_runs,0)+coalesce(extras_2_runs,0))) over(partition by match_id order by match_id,innings asc) as first_innings_target,
lag(count(case when coalesce(lower(extras_type),'') not in ('wide','overstep','other') then runs else null end)) over(partition by match_id order by match_id,innings asc) as first_inning_ball_count
from main_table
where (innings=1 or innings=2) and (first_inn_revised_overs is null and second_inn_revised_overs is null)
group by match_id,innings
),
combined_table as(
select pt.bowler_name,
pt.match_id,
pt.innings,
pt.total_runs_bowler_conceeded,
pt.total_balls_bowled,
pt.total_bowler_wicket,
pt.total_boundary_count,
pt.dot_balls_bowled,
(pt.total_runs_bowler_conceeded)*6.0/nullif((pt.total_balls_bowled),0) as economy_rate,
tt.total_balls_in_match,
round((total_runs_bowler_conceeded*100.0/tt.team_runs),2) as bowler_runs_pct,
tt.total_match_dot_balls,
round((pt.dot_balls_bowled*100.0/nullif(pt.total_balls_bowled,0)),2) as player_dotball_pct,
round((tt.total_match_dot_balls-pt.dot_balls_bowled)*100.0/nullif((tt.total_balls_in_match-pt.total_balls_bowled),0),2) as team_dotball_pct,
tt.early_wickets,
(case when (tt.team_runs-total_runs_bowler_conceeded)*6.0/nullif((tt.total_balls_in_match-pt.total_balls_bowled),0)>9.5 then 'pressure' else 'non pressure' end) as situation
from player_table pt 
inner join team_table tt on pt.match_id=tt.match_id and pt.innings=tt.innings
where pt.total_balls_bowled>=12
),
calculated_table as(
select bowler_name,
sum(case when situation='pressure' then 1 else 0 end) as pressure_situation_count,
count(distinct match_id) as total_matches,
round(sum(case when situation='pressure' then 1 else 0 end)*100.0/count(distinct match_id),2) as pct_pressure_situation, 
round(median(case when situation='pressure' then economy_rate else null end),2) as median_pressure_economy_rate,
round(median(case when situation='non pressure' then economy_rate else null end),2) as median_non_pressure_economy_rate,
round(median(case when situation='pressure' then player_dotball_pct else null end ),2) as median_pressure_dotball_pct,
round(median(case when situation='non pressure' then player_dotball_pct else null end),2) as median_non_pressure_dotball_pct,
round(avg(case when situation='pressure' then total_bowler_wicket else null end),2) as avg_pressure_wickets,
round(avg(case when situation='non pressure' then total_bowler_wicket else null end),2) as avg_non_pressure_wickets
from combined_table
group by bowler_name
having pressure_situation_count>=8
),
final_table as(
select bowler_name,
pressure_situation_count,
median_pressure_economy_rate,
median_non_pressure_economy_rate,
round((100-(median_pressure_economy_rate/nullif(median_non_pressure_economy_rate,0))*100.0),2) as economy_rate_dip_pct,
avg_pressure_wickets,
avg_non_pressure_wickets,
round((median_pressure_dotball_pct-median_non_pressure_dotball_pct),2) as by_how_much_pressure_decreases_dotballs,
round((avg_pressure_wickets-avg_non_pressure_wickets),2) as does_pressure_dip_wicket_taking_ability
from calculated_table
where total_matches>=30 
),
output_table as(
select *  ,
round(median_pressure_economy_rate/nullif(median(median_pressure_economy_rate) over(),0),2) as clutch_index
from final_table
order by clutch_index asc,avg_pressure_wickets desc
),
situation_table as(
SELECT situation, COUNT(*) AS spells_count, ROUND(MEDIAN(economy_rate), 2) AS median_econ, ROUND(MEDIAN(player_dotball_pct), 2) AS median_dot_pct, ROUND(AVG(total_bowler_wicket), 2) AS avg_wickets
FROM combined_table
GROUP BY situation
)
select * from situation_table

""").show()
```
